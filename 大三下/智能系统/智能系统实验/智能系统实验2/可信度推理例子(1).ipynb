# 可信度推理



## 例子1：采矿方法推理

### 1. 知识  

知识库中包含如下规则：

$R_1$:$if$ 地表不允许陷落，矿体埋藏探度一般，  $then$ 采用充填采矿法  $ CF_1(H,E) = 1.0 $     

$R_2$:$if$ 矿石价值很高，$then$ 采用充填采矿法  $CF2(H, E) = 0.75 $  


### 2. 事实

推理数据库中有关于某矿床的如下事实：   

$E_1$：地表不允许陷落，$ CF(E_1)= 0.85 $;   
$E_2$：矿体埋藏探度一般，$ CF(E_2)= 0.90 $   
$E_3$：矿石价值很高，$ Cf(E_3)=1.0; $ 


### 3.可信度推理

记组合证据$E_1$ and $E_2 $的可信度为$ CF(E)$ ，那么

$$
\begin{align} 
CF(E) &= CF(E_1 and E_2) \\&= min\{ CF(E_1), CF(E_2)\} \\
&= min\{ 0.85, 0.90 \} \\&= 0.85 \\
\end{align} 
$$

记规则$R_1$和$ R_2 $的可信度分别为$ CF1(H),CF2(H) $， 那么

$$
\begin{align} 
CF_1(H) &= CF_1(H, E)*max\{0, CF(E)\} \\
&= 1.0 * max\{0, 0.85\}\\ 
&= 1.0 * 0.85\\ 
&= 0.85\\
\end{align} 
$$

$$
\begin{align} 
CF_2(H) &= CF_2(H,E)*max \{ 0,CF(E) \} \\
&= 0.75 * max\{ 0, 1.0 \} \\
&= 0.75 * 1.0 \\ 
&= 0.75\\
\end{align} 
$$

记规则$ R_1 , R_2 $组合规则的可信度为 $ CF_{12}(H) $ ，那么在事实$ E_1, E_2, E_3 $及规则 $ R_1,R_2$ 作用下，结论 $ H $ “采用充填采矿法”的可信度$ CF_{12}(H)的计算可由公式(4),得： 

$$
\begin{align}
CF_1(H) &= 0.8570\\
CF_2(H) &= 0.7570\\
CF_{12}(H) &= CF_1(H) + CF(H) - CF_1(H) * CF_2(H)\\
&= 0.85 + 0.75 -0.85 * 0.75\\
&= 0.96\\
\end{align} 
$$

因此，在事实$E_l,E_2,E_3$及规则$R_1,R_2$的共同作用下，该矿床采用“充填采矿法”开采的可信度为0.96.

## 例子2：预测明天是否下午

### 1. 规则


控制阈值 $\lambda = 0.01 $  
规则$R_1$：if 今天下雨，then 明天下雨；（cf 0.5）  
规则$R_2$：if 今天天气干燥，then 明天天气干燥；（cf 0.5）  
规则$R_3$：if 今天下雨且降雨量小，then 明天天气干燥；（cf 0.6）  
规则$R_4$：if 今天下雨但降雨量小，并且温度低，then 明天天气干燥；（cf 0.7）  
规则$R_5$：if 今天天气干燥，并且温度高，then 明天下雨；（cf 0.65）   
规则$R_6$：if 今天天气干燥，并且温度高，但阴天，then 明天下雨；（cf 0.56）  

### 2. 已知事实

$ E_1 $:今天下雨  $cf = 1$  
$ E_2 $:今天降雨量小  $cf = 0.8$  
$ E_3 $:今天温度低  $cf = 0.9$  

### 3.推理过程


明天天气有两种情况，分别为$H_1$:明天下雨,$H_2$：明天天气干燥  
  
（1）对于事实$E_1$，我们有规则$R_1$：if 今天下雨，then 明天下雨；（cf 0.5） 得：  

$$
\begin{align}
CF_1(H_1,E_1) &= CF(E_1)*cf(R_1)\\
&= 1.0 + 0.5\\
&= 0.5\\
\end{align} 
$$



此时，明天下午的可信度为0.5。

（2）对于事实$ E_2 $:今天降雨量小  $cf = 0.8$ ，由规则$R_3$：if 今天下雨且降雨量小，then 明天天气干燥；（cf 0.6）  得：  
  
$$
\begin{align}
CF_3(H_2,E_1 \cap E_2) &= min[CF(E_1),cf(E_2)] * cf(r_3)\\
&= min[1,0.8] * 0.6\\
&= 0.48\\
\end{align} 
$$

  
即由规则$R_3$可推得结论H2明天的天气干燥的可信度为0.48  


（3）对于事实$ E_3 $:今天温度低  $cf = 0.9$，由规则$R_4$：if 今天下雨但降雨量小，并且温度低，then 明天天气干燥；（cf 0.7）得：

$$
\begin{align}
CF_4(H_2,E_1 \cap E_2 \cap E_3) &= min[CF(E_1),cf(E_2),cf(E_3)] * cf(r_4)\\
&= min[1,0.8,0.9] * 0.7\\
&= 0.56\\
\end{align} 
$$

那么，由规则$R_4$推得结论$H_2$明天干燥的可信度为0.56  


由于规则$R_3,R_4$同时推出结论$H_2$,那么  
$$
\begin{align}
CF(CF_4,CF_3) &= CF_3 + CF_4 * (1-CF_3) \\
&= 0.48  * 0.56 * (1-0.48)\\
&= 0.77\\
\end{align} 
$$

因此，由事实 $E_1,E_2,E_3$ 及 $R_1,R_3,R_4$可推出最终的结果为  
结论H1：明天下雨的可信度为0.5  
结论H2：明天干燥的可信度为0.77
